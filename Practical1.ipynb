```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
picture = cv2.imread("images.jpg")

cv2_imshow(picture)
```

```python
print("Shape:", picture.shape)
```

```python
print("Data Type:", picture.dtype)
```

```python
cropped = picture[150:700, 120:650]

cv2_imshow(cropped)
```

```python
resized = cv2.resize(picture, (400, 500))

cv2_imshow(resized)
```

```python
new_img = picture.copy()

new_img[200, 200] = [0, 255, 0]

cv2_imshow(new_img)
```

```python
gray = cv2.cvtColor(picture, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray)
```

```python
rgb = cv2.cvtColor(picture, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.title("OpenCV (BGR)")
plt.imshow(picture)

plt.subplot(1,2,2)
plt.title("Matplotlib (RGB)")
plt.imshow(rgb)

plt.show()
```

```python
blue = picture[:, :, 0]

plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.show()
```

```python
flip = cv2.flip(picture, 1)

cv2_imshow(flip)
```

```python
rotate = cv2.rotate(picture, cv2.ROTATE_90_CLOCKWISE)

cv2_imshow(rotate)
```

```python
blur = cv2.GaussianBlur(picture, (9,9), 0)

cv2_imshow(blur)
```
